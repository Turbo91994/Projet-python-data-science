Une valeur extrême n'est pas automatiquement une erreur.</p>
</div>

```python
# Identification des valeurs extrêmes par génération
def identifier_extremes(groupe):
    q1 = groupe["bst"].quantile(0.25)
    q3 = groupe["bst"].quantile(0.75)
    iqr = q3 - q1

    return groupe.loc[
        (groupe["bst"] < q1 - 1.5 * iqr) |
        (groupe["bst"] > q3 + 1.5 * iqr)
    ]

extremes = pd.concat(
    [
        identifier_extremes(groupe)
        for _, groupe in df_analyse.groupby("generation")
    ],
    ignore_index=True
)


# Présentation des Pokémon atypiques
display(
    extremes[
        ["nom_fr", "forme", "generation", "bst", "tag"]
    ].sort_values(["generation", "bst"], ascending=[True, False])
)
```

Répartition des catégories spéciales par génération</h2></div><div class="poke-text" style="font:15px/1.75 Arial,sans-serif;color:#24344b;background:#fff;padding:18px 22px;border:1px solid #e1e7ef;border-radius:10px;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Nous étudions la répartition des Pokémon légendaires, fabuleux, Ultra-Chimères et Paradoxes selon leur génération d'introduction, afin de mieux comprendre la composition des générations.</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Ici, une observation représente une <strong>espèce</strong>, classée selon sa génération d'introduction. Cela permet de décrire la composition sans compter plusieurs fois ses formes. Les catégories peuvent se chevaucher.</p>
</div><p style="font:12px Arial,sans-serif;margin:12px 0 20px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><a href="#sommaire" style="color:#2456a6;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Retour au sommaire</a></p>

```python


# Génération d'introduction de chaque espèce
especes = forms[
    ["species_id", "species_generation"]
].drop_duplicates()

# Catégories spéciales associées aux espèces
categories = ["Légendaire", "Fabuleux", "Ultra-Chimère", "Paradoxe"]

df_cat_especes = base[["species_id", "tag"]].drop_duplicates().copy()
df_cat_especes["tag"] = df_cat_especes["tag"].fillna("").str.split(";")
df_cat_especes = df_cat_especes.explode("tag")
df_cat_especes["tag"] = df_cat_especes["tag"].str.strip()

df_cat_especes = df_cat_especes[
    df_cat_especes["tag"].isin(categories)
]

# Association avec la génération de l'espèce
df_cat_especes = df_cat_especes.merge(
    especes, on="species_id", how="left", validate="many_to_one"
).drop_duplicates(["species_id", "tag"])

# Nombre d'espèces par génération et catégorie
repartition = pd.crosstab(
    df_cat_especes["species_generation"],
    df_cat_especes["tag"]
).reindex(index=range(1, 10), columns=categories, fill_value=0)

plt.figure(figsize=(10, 5))
sns.heatmap(repartition, annot=True, fmt="d", cmap="Blues")

plt.title("Répartition des espèces spéciales par génération")
plt.xlabel("Catégorie")
plt.ylabel("Génération d'introduction de l'espèce")
plt.tight_layout()
plt.show()





```

BST moyen selon la catégorie</h2></div><div class="poke-text" style="font:15px/1.75 Arial,sans-serif;color:#24344b;background:#fff;padding:18px 22px;border:1px solid #e1e7ef;border-radius:10px;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Chaque case indique le <strong>BST moyen</strong> et le nombre <code style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">n</code> d'observations variété-génération. La couleur représente la moyenne ; une case grise signifie que la catégorie est absente de cette génération, et non que sa moyenne vaut zéro.</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Les catégories se chevauchent : une observation " Légendaire; Paradoxe " contribue aux deux groupes. Les effectifs ne représentent donc pas des espèces distinctes et ne doivent pas être additionnés.</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Nous revenons aux couples variété-génération pour comparer les statistiques : les effectifs de cette grille ne sont donc pas ceux du tableau d'espèces précédent.</p>
</div><p style="font:12px Arial,sans-serif;margin:12px 0 20px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><a href="#sommaire" style="color:#2456a6;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Retour au sommaire</a></p>

```python
# Une observation par variété et génération, comptée dans chacun de ses tags.
categories = ["Légendaire", "Fabuleux", "Ultra-Chimère", "Paradoxe"]
df_cat = df_analyse.copy()
df_cat["tag"] = df_cat["tag"].fillna("").str.split(";")
df_cat = df_cat.explode("tag")
df_cat["tag"] = df_cat["tag"].str.strip()
df_cat = df_cat[df_cat["tag"].isin(categories)]

bst_cat = df_cat.groupby(["generation", "tag"])["bst"].agg(["mean", "count"]).reset_index()
moyennes = bst_cat.pivot(index="tag", columns="generation", values="mean")
# float permet à Seaborn de lire les valeurs manquantes comme NaN.
moyennes = moyennes.reindex(index=categories, columns=range(1, 10)).astype(float)
effectifs = bst_cat.pivot(index="tag", columns="generation", values="count")

# Préparer les deux lignes de texte affichées dans chaque case.
annotations = moyennes.astype(object).copy()
for categorie in categories:
    for generation in moyennes.columns:
        moyenne = moyennes.loc[categorie, generation]
        if pd.notna(moyenne):
            effectif = int(effectifs.loc[categorie, generation])
            annotations.loc[categorie, generation] = f"{moyenne:.0f}\nn = {effectif}"
        else:
            annotations.loc[categorie, generation] = ""

fig, ax = plt.subplots(figsize=(13, 5))
ax.set_facecolor("#edf1f4")
sns.heatmap(
    moyennes, annot=annotations, fmt="", cmap="YlGnBu",
    linewidths=3, linecolor="white", annot_kws={"fontsize": 11},
    cbar_kws={"label": "BST moyen", "shrink": 0.8}, ax=ax
)
# Un tiret distingue explicitement les catégories absentes.
for ligne, categorie in enumerate(categories):
    for colonne, generation in enumerate(moyennes.columns):
        if pd.isna(moyennes.loc[categorie, generation]):
            ax.text(colonne + 0.5, ligne + 0.5, "—", ha="center", va="center", color="#82909b")

fig.suptitle("BST moyen par catégorie et génération", x=0.02, ha="left",
             fontsize=18, fontweight="bold", color="#183b4e")
ax.set_title("Moyenne et effectif de chaque groupe", loc="left", fontsize=11, color="#526777", pad=15)
ax.set_xlabel("Génération d’introduction", labelpad=12)
ax.set_ylabel("")
ax.tick_params(axis="both", length=0, labelsize=11)
ax.tick_params(axis="y", rotation=0)
fig.text(0.02, 0.02, "n : observations variété–génération  ·  Gris : catégorie absente  ·  Catégories non exclusives",
         fontsize=10, color="#526777")
fig.tight_layout(rect=(0, 0.07, 1, 0.94))
plt.show()
```

<div id="specialisation" style="scroll-margin-top:20px;background:#eef3fb;border-left:6px solid #ffcb05;border-radius:8px;padding:18px 22px;margin:26px 0 12px;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><h2 style="font: bold 21px/1.35 Arial,sans-serif;color:#173f80;margin:0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">10.7. Évolution de la spécialisation</h2></div><div class="poke-text" style="font:15px/1.75 Arial,sans-serif;color:#24344b;background:#fff;padding:18px 22px;border:1px solid #e1e7ef;border-radius:10px;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">La spécialisation mesure le déséquilibre entre les six statistiques d'un Pokémon. Un Pokémon spécialisé possède certaines statistiques nettement plus élevées que d'autres.</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">On la mesure par le coefficient de variation :</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">$$
CV = \frac{\text{Écart-type des six statistiques}}{\text{Moyenne des six statistiques}}
$$</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Plus le CV est élevé, plus le Pokémon est spécialisé. Nous étudions son évolution pour déterminer si les générations récentes présentent des profils statistiques plus déséquilibrés.</p>
</div><p style="font:12px Arial,sans-serif;margin:12px 0 20px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><a href="#sommaire" style="color:#2456a6;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Retour au sommaire</a></p>

```python
# Les six statistiques de base
stats = [
    "pv", "attaque", "defense",
    "attaque_speciale", "defense_speciale", "vitesse"
]

# Calcul du coefficient de variation pour chaque Pokémon
df_analyse["specialisation"] = (
    df_analyse[stats].std(axis=1, ddof=0)
    / df_analyse[stats].mean(axis=1)
)

# Visualisation de la distribution par génération
plt.figure(figsize=(11, 5))

sns.boxplot(
    data=df_analyse,
    x="generation",
    y="specialisation",
    color="lightsteelblue"
)

plt.title("Évolution de la spécialisation par génération")
plt.xlabel("Génération")
plt.ylabel("Coefficient de variation")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

<div id="extremes-specialisation" style="scroll-margin-top:20px;background:#eef3fb;border-left:6px solid #ffcb05;border-radius:8px;padding:18px 22px;margin:26px 0 12px;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><h2 style="font: bold 21px/1.35 Arial,sans-serif;color:#173f80;margin:0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">10.7.1. Les profils atypiques</h2></div><div class="poke-text" style="font:15px/1.75 Arial,sans-serif;color:#24344b;background:#fff;padding:18px 22px;border:1px solid #e1e7ef;border-radius:10px;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Comme pour le BST, nous utilisons les bornes <strong>Q1 - 1,5 x IQR</strong> et <strong>Q3 + 1,5 x IQR</strong>, calculées séparément dans chaque génération, avec <code style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">IQR = Q3 - Q1</code>.</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Une valeur extrême <strong>haute</strong> indique un profil particulièrement déséquilibré ; une valeur extrême <strong>basse</strong>, un profil particulièrement équilibré par rapport aux autres observations de sa génération. Ce repérage décrit les points au-delà des moustaches du graphique précédent : il ne signale pas automatiquement une erreur dans les données et ne justifie aucune suppression.</p>
<p style=";box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Nous conservons la même unité que dans cette section : un couple variété-génération. Le tableau récapitulatif donne aussi la proportion d'observations atypiques, car les générations ont des effectifs différents.</p>
</div><p style="font:12px Arial,sans-serif;margin:12px 0 20px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><a href="#sommaire" style="color:#2456a6;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Retour au sommaire</a></p><div class="poke-art" style="text-align:center;background:#fff;border:1px solid #e1e7ef;border-radius:12px;padding:16px;margin:14px 0;box-sizing:border-box;width:100%;max-width:100%;min-width:0;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><img alt="Caratroc" height="160" src="https://raw.githubusercontent.com/PokeAPI/sprites/master/sprites/pokemon/other/official-artwork/213.png" style="display:inline-block;width:160px;height:160px;object-fit:contain;margin:0 12px;box-sizing:border-box;max-width:calc(100% - 24px);height:auto;" width="160"/><p style="text-align:center;font:13px Arial,sans-serif;color:#52627b;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Caratroc : un profil déséquilibré que ses six statistiques permettent de comprendre.</p></div>

```python
# La même règle que pour le BST, appliquée au coefficient de variation.
resume_specialisation = []
groupes_extremes = []
for generation, groupe in df_analyse.groupby("generation"):
    q1 = groupe["specialisation"].quantile(0.25)
    q3 = groupe["specialisation"].quantile(0.75)
    iqr = q3 - q1
    borne_basse = q1 - 1.5 * iqr
    borne_haute = q3 + 1.5 * iqr

    bas = groupe["specialisation"] < borne_basse
    hauts = groupe["specialisation"] > borne_haute
    atypiques = groupe.loc[bas | hauts].copy()
    atypiques["position"] = "Haute"
    atypiques.loc[atypiques["specialisation"] < borne_basse, "position"] = "Basse"
    groupes_extremes.append(atypiques)

    resume_specialisation.append({
        "generation": generation,
        "observations": len(groupe),
        "borne_basse": borne_basse,
        "borne_haute": borne_haute,
        "extremes_bas": bas.sum(),
        "extremes_hauts": hauts.sum(),
        "part_extremes_pct": 100 * len(atypiques) / len(groupe)
    })

resume_specialisation = pd.DataFrame(resume_specialisation)
extremes_specialisation = pd.concat(groupes_extremes, ignore_index=True)
display(resume_specialisation.round(3))
```

```python
# Les six statistiques permettent d’expliquer chaque profil atypique.
colonnes = ["generation", "nom_fr", "forme", "position", "specialisation", "bst"] + stats
profils_extremes = extremes_specialisation[colonnes].sort_values(
    ["generation", "specialisation"], ascending=[True, False]
)
# Afficher tous les cas repérés, sans tronquer la liste.
with pd.option_context("display.max_rows", None, "display.max_columns", None):
    display(profils_extremes.round(3))
```

Méga-Évolutions afin d'évaluer leur contribution aux
différences observées entre générations.</p>
</div><p style="font:12px Arial,sans-serif;margin:12px 0 20px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;"><a href="#sommaire" style="color:#2456a6;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;word-break:normal;white-space:normal;">Retour au sommaire</a></p>

```python
# Identification des Méga-Évolutions
mega = df_analyse["is_mega"].astype(str).str.lower().isin(["true", "1"])

print("Nombre de Méga-Évolutions :", mega.sum())

# Moyennes avec et sans Méga-Évolutions
bst_tous = df_analyse.groupby("generation")["bst"].mean()

bst_sans_mega = (
    df_analyse.loc[~mega]
    .groupby("generation")["bst"]
    .mean()
)

# Visualisation
plt.figure(figsize=(10, 5))

plt.plot(bst_tous.index, bst_tous,
         marker="o", label="Toutes les variétés")

plt.plot(bst_sans_mega.index, bst_sans_mega,
         marker="s", linestyle="--",
         label="Sans Méga-Évolutions")

plt.title("Influence des Méga-Évolutions sur le BST moyen")
plt.xlabel("Génération")
plt.ylabel("BST moyen")
plt.xticks(range(1, 10))
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
